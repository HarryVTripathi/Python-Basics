### Two tier sort

```python
data = [('hi', 1), ('how', 1), ('are', 2), ('you', 1), ('don', 1), ('t', 1), ('do', 1), ('it', 1), ('ok', 1)]
result = sorted(data, key=lambda x: (-x[1], x[0]))
```

```python
result
```

### Sorting with comparator

In Python 3, `sorted()` **does not** directly accept a comparator function like it did in Python 2 (with the `cmp` parameter).

Instead, you must use a `key` function, which maps each element to a value used for sorting.

However,we you **really want** to use a comparator (custom comparison between two elementwe you can **wra** your comparator using `functools.cmp_to`_key cool one-liner too!) 😎

```python
import functools
```

```python
def compare(a, b):
    if a[1] != b[1]:
        return b[1] - a[1]
    return (a[0] > b[0]) - (a[0] < b[0])


data = [('hi', 1), ('how', 1), ('are', 2), ('you', 1), ('don', 1), ('t', 1), ('do', 1), ('it', 1), ('ok', 1)]
result = sorted(data, key=functools.cmp_to_key(compare))
```

```python
result
```

### Sorting class objects

```python
from datetime import date

class Book:
    def __init__(self, title, published):
        self.title = title
        self.published = published  # assume datetime.date

    def __repr__(self):
        return f"Book({self.title!r}, {self.published})"

    def __lt__(self, other):
        # Recent books should come first (so we "invert" the date comparison)
        if self.published != other.published:
            return self.published > other.published  # newer first

        return self.title < other.title  # alphabetical
```

```python
books = [
    Book("Alpha", date(2023, 5, 1)),
    Book("Beta", date(2024, 1, 10)),
    Book("Gamma", date(2023, 5, 1)),
    Book("Delta", date(2022, 3, 15))
]

sorted_books = sorted(books)
print(sorted_books)
```
