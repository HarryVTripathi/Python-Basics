```python
from dataclasses import dataclass, field
from typing import Optional
```

```python
@dataclass
class Config:
    config_1: Optional[str] = field(default=None, init=False)
    config_2: Optional[str] = field(default=None, init=False)
    _instance: Optional["Config"] = field(default=None, init=False, repr=False)

    def __post_init__(self):
        print("Initializing")
        self.config_1 = "some_value"
        self.config_2 = "some_value"

        if self._instance is None:
            Config._instance = self

    @classmethod
    def get_instance(cls) -> "Config":
        if cls._instance is None:
            print("Creating new instance.")
            cls()
        return cls._instance
```

```python
config = Config.get_instance()
```
